# 🎬 Netflix Recommendation System — Full ML Pipeline
**Dataset:** netflix_titles.csv (6234 records, 12 features)  
**Goal:** Build a content-based recommendation system using NLP + ML  
**Pipeline:** EDA → Preprocessing → Feature Engineering → Model Training → Evaluation → PKL Export

---
## Architecture Overview
```
Raw CSV → EDA (Before/After) → Text Preprocessing → TF-IDF Vectorization
       → 5 ML Models → Best Model (AUC/ROC) → Confusion Matrix → PKL Save
```

**Input Parameters:**
- `title` — The Netflix title to base recommendations on
- `type` — Movie or TV Show filter
- `listed_in` — Genre tags (multi-label)
- `description` — Free-text synopsis (main NLP feature)
- `cast`, `director`, `country`, `rating` — Content metadata
- `release_year` — Temporal feature
- `duration` — Runtime (minutes for movies, seasons for shows)

**Why Content-Based Filtering?**  
Netflix titles have rich metadata but no user interaction data in this dataset.  
Content-based filtering uses item features (genres, description, cast) to find similar items — no user history needed.

## CELL 4 — Data Preprocessing (Cleaning Pipeline)

````python
# ── PREPROCESSING PIPELINE ───────────────────────────────────────────────
print('=' * 60)
print('PREPROCESSING PIPELINE')
print('=' * 60)

# STEP 1: Fill missing values
print('\n[STEP 1] Filling Missing Values...')
df['director'] = df['director'].fillna('Unknown Director')
df['cast'] = df['cast'].fillna('Unknown Cast')
df['country'] = df['country'].fillna('Unknown Country')
df['rating'] = df['rating'].fillna(df['rating'].mode()[0])  # mode for categorical
df['date_added'] = df['date_added'].fillna('Unknown Date')
print(f'   Missing after fill: {df.isnull().sum().sum()} nulls')

# STEP 2: Parse date_added → year_added
print('\n[STEP 2] Parsing date_added → year_added...')
df['year_added'] = pd.to_datetime(df['date_added'], errors='coerce').dt.year
df['year_added'] = df['year_added'].fillna(df['release_year'])  # fallback
df['year_added'] = df['year_added'].astype(int)

# STEP 3: Parse duration → numeric
print('\n[STEP 3] Parsing duration to numeric...')
df['duration_val'] = df['duration'].str.extract(r'(\d+)').astype(float)
df['duration_unit'] = df['duration'].str.extract(r'(min|Season)')
# Normalize: seasons * 10 to put on similar scale as minutes (heuristic)
df['duration_numeric'] = np.where(df['duration_unit'] == 'Season',
                                   df['duration_val'] * 10,
                                   df['duration_val'])

# STEP 4: Clip outliers on release_year
print('\n[STEP 4] Clipping outliers on release_year...')
col = 'release_year'
Q1, Q3 = df[col].quantile(0.25), df[col].quantile(0.75)
IQR = Q3 - Q1
df[col] = np.clip(df[col], Q1 - 1.5*IQR, Q3 + 1.5*IQR)

# STEP 5: Standardize text columns
print('\n[STEP 5] Standardizing text columns...')
for col in ['title', 'director', 'cast', 'country', 'listed_in', 'description']:
    df[col] = df[col].astype(str).str.strip()

# STEP 6: Extract primary genre
print('\n[STEP 6] Extracting primary_genre...')
df['primary_genre'] = df['listed_in'].str.split(',').str[0].str.strip()

# STEP 7: Derive content_era from release_year
print('\n[STEP 7] Deriving content_era...')
def get_era(year):
    if year < 1990: return 'Classic'
    elif year < 2000: return '90s'
    elif year < 2010: return '2000s'
    elif year < 2015: return 'Early 2010s'
    else: return 'Recent'
df['content_era'] = df['release_year'].apply(get_era)

print('\n✅ Preprocessing complete!')
print(f'Final shape: {df.shape}')
df.head(3)
````

## CELL 7 — NLP Feature Engineering (Soup / TF-IDF)

````python
# ── NLP TEXT PREPROCESSING ───────────────────────────────────────────────
print('=' * 60)
print('NLP FEATURE ENGINEERING')
print('=' * 60)

# Load cleaned data (ensures fresh start from saved state)
df = pd.read_csv('netflix_cleaned.csv')

stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def clean_text(text):
    """Clean and normalize a text string."""
    text = str(text).lower()
    text = re.sub(r'[^a-z\s]', '', text)          # remove punctuation/numbers
    tokens = text.split()
    tokens = [lemmatizer.lemmatize(t) for t in tokens if t not in stop_words]
    return ' '.join(tokens)

def build_soup(row):
    """
    Build a 'content soup' combining all text features.
    Weights: description (3x), genre (2x), cast (2x), director (1x), country (1x)
    WHY: Higher weight on description captures semantic meaning.
          Genre and cast weight ensures genre/actor-based similarity.
    """
    description = clean_text(row['description']) + ' ' + clean_text(row['description']) + ' ' + clean_text(row['description'])
    genres = clean_text(row['listed_in']) + ' ' + clean_text(row['listed_in'])
    cast = clean_text(row['cast'].replace(',', ' ')) + ' ' + clean_text(row['cast'].replace(',', ' '))
    director = clean_text(row['director'])
    country = clean_text(row['country'].split(',')[0])  # primary country
    era = clean_text(row['content_era'])
    type_tag = clean_text(row['type'])
    return f'{description} {genres} {cast} {director} {country} {era} {type_tag}'

print('Building content soup (weighted text combination)...')
df['soup'] = df.apply(build_soup, axis=1)

print(f'\nSample soup for "{df["title"].iloc[0]}":')
print(df['soup'].iloc[0][:200], '...')
print(f'\n✅ soup column created for {len(df)} titles')
````

````python
# ── TF-IDF VECTORIZATION ─────────────────────────────────────────────────
print('=' * 60)
print('TF-IDF VECTORIZATION')
print('=' * 60)
print("""
WHY TF-IDF?
  TF (Term Frequency)   : How often a word appears in a document
  IDF (Inverse Document Frequency): How unique the word is across all documents
  TF-IDF = TF × IDF  → Words that are specific to a title get higher weight
  This avoids common words (like 'film', 'story') dominating recommendations.
""")

tfidf = TfidfVectorizer(
    max_features=10000,      # top 10k vocabulary
    ngram_range=(1, 2),      # unigrams + bigrams (e.g., 'action thriller')
    min_df=2,                # ignore terms appearing in fewer than 2 docs
    sublinear_tf=True        # use log(tf) to dampen high-frequency terms
)

tfidf_matrix = tfidf.fit_transform(df['soup'])

print(f'TF-IDF Matrix Shape: {tfidf_matrix.shape}')
print(f'  → {tfidf_matrix.shape[0]} titles × {tfidf_matrix.shape[1]} features')
print(f'  Sparsity: {100 * (1 - tfidf_matrix.nnz / (tfidf_matrix.shape[0] * tfidf_matrix.shape[1])):.1f}%')

# Build title → index map
indices = pd.Series(df.index, index=df['title'].str.lower()).drop_duplicates()

print('\n✅ TF-IDF vectorization complete!')
````

## CELL 8 — Cosine Similarity (Core Recommendation Engine)

````python
# ── COSINE SIMILARITY ────────────────────────────────────────────────────
print('=' * 60)
print('COSINE SIMILARITY — CONTENT-BASED FILTERING')
print('=' * 60)
print("""
WHY COSINE SIMILARITY?
  Measures the angle between two TF-IDF vectors.
  Score = 1.0 → identical content
  Score = 0.0 → completely different
  Works best for text data because it's length-agnostic.
  Formula: cos(θ) = (A·B) / (||A|| × ||B||)
""")

# Compute cosine similarity matrix
cosine_sim = linear_kernel(tfidf_matrix, tfidf_matrix)

print(f'Similarity Matrix Shape: {cosine_sim.shape}')
print(f'Sample diagonal (self-similarity): {cosine_sim[0,0]:.4f} (should be 1.0)')

def get_recommendations(title, n=10, content_type=None):
    """
    Get top-N content recommendations for a given Netflix title.
    
    Parameters:
    -----------
    title        : str — exact title (case-insensitive)
    n            : int — number of recommendations (default 10)
    content_type : str — filter by 'Movie' or 'TV Show' (optional)
    
    Returns:
    --------
    DataFrame with recommended titles and similarity scores
    """
    title_lower = title.lower()
    
    if title_lower not in indices:
        # Fuzzy match suggestion
        candidates = [t for t in indices.index if title_lower[:5] in t]
        print(f'⚠️ "{title}" not found. Did you mean: {candidates[:3]}?')
        return None
    
    idx = indices[title_lower]
    sim_scores = list(enumerate(cosine_sim[idx]))
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)
    sim_scores = [s for s in sim_scores if s[0] != idx]  # exclude self
    
    top_indices = [s[0] for s in sim_scores[:50]]  # top 50 candidates
    recs = df.iloc[top_indices][['title', 'type', 'listed_in', 'release_year',
                                  'rating', 'country', 'description']].copy()
    recs['similarity_score'] = [round(s[1], 4) for s in sim_scores[:50]]
    
    if content_type:
        recs = recs[recs['type'] == content_type]
    
    return recs.head(n)

# Test recommendation
print('\n--- TEST RECOMMENDATION ---')
sample_title = df['title'].iloc[100]
print(f'Input title: "{sample_title}"')
recs = get_recommendations(sample_title, n=5)
if recs is not None:
    print(recs[['title', 'type', 'similarity_score', 'listed_in']].to_string(index=False))
````

## CELL 9 — ML Task Definition: Genre Classification

````python
# ── ML TASK: GENRE CLASSIFICATION ────────────────────────────────────────
print('=' * 60)
print('ML TASK: CONTENT-TYPE CLASSIFICATION')
print('=' * 60)
print("""
WHY A CLASSIFICATION MODEL ALONGSIDE RECOMMENDATION?
  The recommender is similarity-based (unsupervised).
  We add a supervised ML layer to:
    1. Predict content_type (Movie vs TV Show) from features — validates
       which features best describe content identity
    2. The best classifier is used to rank/filter recommendations
    3. AUC/ROC on binary task gives interpretable model comparison

INPUT PARAMETERS:
  X: TF-IDF soup features (text-based) + numeric features
  y: type (Movie=0, TV Show=1) — binary classification
""")

# Build feature matrix for ML
from sklearn.preprocessing import LabelEncoder
from scipy.sparse import hstack, csr_matrix

# Encode target
le = LabelEncoder()
y = le.fit_transform(df['type'])   # Movie=0, TV Show=1
print(f'Target distribution: {dict(zip(le.classes_, np.bincount(y)))}')

# Numeric features
scaler = MinMaxScaler()
num_feats = scaler.fit_transform(df[['release_year', 'duration_numeric', 'year_added']])
num_feats_sparse = csr_matrix(num_feats)

# Combined feature matrix: TF-IDF + numeric
X = hstack([tfidf_matrix, num_feats_sparse])
print(f'\nFeature matrix shape: {X.shape}')
print(f'  {tfidf_matrix.shape[1]} TF-IDF features + 3 numeric features')

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=SEED, stratify=y
)
print(f'\nTrain: {X_train.shape[0]} | Test: {X_test.shape[0]}')
print('✅ Data ready for ML training!')
````

## CELL 10 — Train 5 ML Models & Compare

````python
# ── 5 ML MODELS ──────────────────────────────────────────────────────────
print('=' * 60)
print('TRAINING 5 ML MODELS')
print('=' * 60)

models = {
    'Logistic Regression': LogisticRegression(
        max_iter=1000, C=1.0, solver='lbfgs', random_state=SEED),
    
    'Random Forest': RandomForestClassifier(
        n_estimators=100, max_depth=15, random_state=SEED, n_jobs=-1),
    
    'Gradient Boosting': GradientBoostingClassifier(
        n_estimators=100, learning_rate=0.1, max_depth=5, random_state=SEED),
    
    'SVM (Linear)': SVC(
        kernel='linear', C=1.0, probability=True, random_state=SEED),
    
    'KNN (Nearest Neighbors)': NearestNeighbors(n_neighbors=5, metric='cosine')
}

results = {}

for name, model in models.items():
    print(f'\n▶ Training: {name}...')
    
    if name == 'KNN (Nearest Neighbors)':
        # KNN is used for unsupervised recommendation; skip classification eval
        model.fit(X_train)
        results[name] = {'accuracy': 'N/A (unsupervised)', 'auc': 'N/A',
                         'f1': 'N/A', 'model': model}
        print(f'   KNN fitted for recommendation use (unsupervised)')
        continue
    
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:, 1]
    
    acc = accuracy_score(y_test, y_pred)
    f1  = f1_score(y_test, y_pred, average='weighted')
    auc_score = roc_auc_score(y_test, y_prob)
    
    results[name] = {
        'accuracy': round(acc, 4),
        'f1': round(f1, 4),
        'auc': round(auc_score, 4),
        'model': model,
        'y_prob': y_prob
    }
    print(f'   Accuracy: {acc:.4f} | F1: {f1:.4f} | AUC: {auc_score:.4f}')

print('\n✅ All models trained!')
````

````python
# ── OVERFITTING / UNDERFITTING CHECK ─────────────────────────────────────
print('=' * 60)
print('OVERFITTING / UNDERFITTING ANALYSIS')
print('=' * 60)
print("""
DIAGNOSIS GUIDE:
  Train Acc >> Test Acc  → OVERFITTING  (model memorized training data)
  Both low              → UNDERFITTING (model too simple)
  Train ≈ Test, both high → GOOD FIT
""")

train_acc = accuracy_score(y_train, best_model.predict(X_train))
test_acc  = accuracy_score(y_test,  y_pred_best)
gap = train_acc - test_acc

print(f'  Train Accuracy : {train_acc:.4f}')
print(f'  Test  Accuracy : {test_acc:.4f}')
print(f'  Generalization Gap: {gap:.4f}')

if gap > 0.05:
    print('\n⚠️  OVERFITTING DETECTED — gap > 5%')
    print('   Remedies: increase regularization (C param), reduce max_features, add dropout')
elif test_acc < 0.75:
    print('\n⚠️  UNDERFITTING DETECTED — test accuracy < 75%')
    print('   Remedies: try more complex model, increase features, reduce regularization')
else:
    print('\n✅ GOOD FIT — No significant overfitting or underfitting')

# Cross-validation for robust evaluation
print('\n--- 5-Fold Cross-Validation ---')
cv_scores = cross_val_score(best_model, X, y, cv=5, scoring='roc_auc', n_jobs=-1)
print(f'  CV AUC Scores: {[round(s,4) for s in cv_scores]}')
print(f'  Mean CV AUC : {cv_scores.mean():.4f} ± {cv_scores.std():.4f}')

if cv_scores.std() < 0.02:
    print('\n✅ Low variance across folds — model is stable!')
else:
    print('\n⚠️  High variance — consider tuning or more data')
````

## CELL 14 — Load & Run Inference (Production Test)

````python
# ── PRODUCTION INFERENCE TEST ─────────────────────────────────────────────
print('=' * 60)
print('PRODUCTION INFERENCE — Load from PKL & Recommend')
print('=' * 60)

# Reload everything from disk
tfidf_loaded     = joblib.load('models/tfidf_vectorizer.pkl')
cos_sim_loaded   = joblib.load('models/cosine_similarity.pkl')
indices_loaded   = joblib.load('models/title_indices.pkl')
df_loaded        = joblib.load('models/netflix_df.pkl')
scaler_loaded    = joblib.load('models/minmax_scaler.pkl')
le_loaded        = joblib.load('models/label_encoder.pkl')

def recommend(title, n=10, filter_type=None):
    """
    Full inference pipeline using loaded PKL artifacts.
    
    Parameters:
    -----------
    title       : str  - Netflix title to find recommendations for
    n           : int  - Number of recommendations to return
    filter_type : str  - 'Movie' or 'TV Show' (optional)
    
    Returns:
    --------
    DataFrame of recommended titles sorted by similarity score
    """
    t = title.lower()
    if t not in indices_loaded:
        print(f'❌ Title "{title}" not found in database.')
        return None
    
    idx = indices_loaded[t]
    scores = sorted(enumerate(cos_sim_loaded[idx]), key=lambda x: x[1], reverse=True)
    scores = [s for s in scores[1:51]]  # top 50, skip self
    
    top_idx = [s[0] for s in scores]
    out = df_loaded.iloc[top_idx][['title', 'type', 'listed_in', 'release_year', 'rating']].copy()
    out['similarity'] = [round(s[1], 4) for s in scores]
    
    if filter_type:
        out = out[out['type'] == filter_type]
    
    return out.head(n)

# Demo
print('\n🎬 RECOMMENDATIONS DEMO:')
test_titles = [
    df_loaded['title'].iloc[50],
    df_loaded[df_loaded['type']=='TV Show']['title'].iloc[10]
]

for t in test_titles:
    print(f'\n▶ Input: "{t}"')
    r = recommend(t, n=5)
    if r is not None:
        print(r[['title', 'type', 'similarity', 'listed_in']].to_string(index=False))
````
